```python
#Vishal S 24BAD130
#SCENARIO 3: Housing Dataset Investigate a housing dataset for missing features and relationships between house attributes
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print("Vishal S 24BAD130")
df = pd.read_csv("Housing.csv")
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print("Vishal S 24BAD130")
plt.scatter(df['area'], df['price'])
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
