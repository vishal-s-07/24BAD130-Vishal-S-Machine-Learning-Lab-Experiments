```python
#Vishal S 24BAD130
#SCENARIO 1: E-commerce Sales Data Analyze a sample e-commerce dataset to understand product categories, sales trends, and missing information.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
print("Vishal S 24BAD130")
df=pd.read_csv("data.csv")
```

```python
df['Sales'] = df['Quantity'] * df['UnitPrice']
```

```python
sales_per_product = df.groupby('Description')['Sales'].sum().head(10)
```

```python
sales_per_product.plot(kind='bar')
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.title("Sales per Product")
plt.show()
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce', format='mixed')
daily_sales = df.groupby(df['InvoiceDate'].dt.date)['Sales'].sum()
daily_sales.plot()
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.title("Daily Sales Trend")
plt.show()
```
